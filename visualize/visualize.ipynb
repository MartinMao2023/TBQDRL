```python
import jax
import jax.numpy as jnp
import numpy as np

import os
# Tell MuJoCo to use hardware-accelerated headless rendering
os.environ['MUJOCO_GL'] = 'egl'


import mujoco
from brax import envs
# from IPython.display import HTML, display
import mediapy as media
from robots import go2
from task_wrappers.go2_wrapper import Go2Wrapper
```

```python
seed = 42
# seed = 4444
loop_random_key = jax.random.PRNGKey(seed)
loop_random_key, subkey = jax.random.split(loop_random_key)

# 1. Ensure your environment is imported and registered
# import go2_env  <-- Uncomment this if your env class is in a separate file

# 2. Create the environment
# We don't need AutoResetWrapper here because we just want a single clean rollout
env = envs.create('go2_mjx', episode_length=4096, backend="mjx", auto_reset=True)
env = Go2Wrapper(env)
```

```python
jit_reset = jax.jit(env.reset)
jit_step = jax.jit(env.step)
state = jit_reset(subkey)
```

```python
initial_shape = []
for i in jax.tree.leaves(state):
    initial_shape.append(i.shape)
```

```python
zero_action = jnp.zeros(env.action_size)
state, _ = jit_step(state, zero_action)
```

```python
new_shape = []
for i in jax.tree.leaves(state):
    new_shape.append(i.shape)
```

```python
for i, j in zip(initial_shape, new_shape):
    if i != j:
        print(i, j)
```

```python
trajectory = [state.env_state.pipeline_state]
episode_length = 300
key = loop_random_key
```

```python
print(f"Simulating {episode_length} steps on GPU...")
for _ in range(episode_length):
    key, subkey = jax.random.split(key)
    candidate_action_noise = 0.25 * jax.random.normal(subkey, zero_action.shape)
    action = jnp.clip(candidate_action_noise, -1.0, 1.0)
    state, _ = jit_step(state, action)
    trajectory.append(state.env_state.pipeline_state)

print("Rendering video on CPU...")

mj_model = env.unwrapped._mj_model
mj_data = mujoco.MjData(mj_model)
renderer = mujoco.Renderer(mj_model, height=480, width=640)

frames = []

# 4. Loop through the GPU trajectory and render it
for pipeline_state in trajectory:
    # Pull JAX arrays from the GPU back to standard Numpy CPU arrays
    mj_data.qpos = np.array(pipeline_state.qpos)
    mj_data.qvel = np.array(pipeline_state.qvel)
    
    # Run forward kinematics to update the 3D geometry of the robot based on qpos
    mujoco.mj_forward(mj_model, mj_data)
    
    # Render the frame
    renderer.update_scene(mj_data)
    frames.append(renderer.render())

# 5. Display the video in the Jupyter Notebook (Playback at 50Hz)
media.show_video(frames, fps=1.0/env.dt)
```
